# ข้อ 3 ระบุข้อมูลสูญหาย

```python
filtered_combined_df.isnull().sum()
```

```python
# Group by 'tele_station_oldcode' and sum nulls in 'rainfall_value' and 'tele_station_id'
missing_per_station = filtered_combined_df.groupby('tele_station_oldcode')[['rainfall_value', 'tele_station_id']].apply(lambda x: x.isnull().sum())

# Filter to show only stations with missing rainfall values in either 'rainfall_value' or 'tele_station_id'
# แก้ไข: ใช้ missing_per_station แทน missingl_station และกรองสถานีที่มีค่าสูญหายมากกว่า 0 ในคอลัมน์ใดคอลัมน์หนึ่ง
missing_per_station_filtered = missing_per_station[(missing_per_station['rainfall_value'] > 0) | (missing_per_station['tele_station_id'] > 0)]

if not missing_per_station_filtered.empty:
    print("จำนวนข้อมูลที่สูญหายไปในแต่ละสถานี (ในคอลัมน์ 'rainfall_value' หรือ 'tele_station_id'):")
    display(missing_per_station_filtered.sort_values(by=['rainfall_value', 'tele_station_id'], ascending=False))
else:
    print("ไม่พบข้อมูลที่สูญหายไปในสถานีใดๆ ในคอลัมน์ 'rainfall_value' หรือ 'tele_station_id'")
```

#### คำนวณสัดส่วนข้อมูลที่สูญหายไปของแต่ละสถานี

```python
# 1. คำนวณจำนวนข้อมูลทั้งหมดของแต่ละสถานี
total_records_per_station = filtered_combined_df.groupby('tele_station_oldcode').size().rename('total_records')

# 2. รวมข้อมูลที่หายไปกับจำนวนข้อมูลทั้งหมด
# ใช้ .reset_index() เพื่อให้ 'tele_station_oldcode' เป็นคอลัมน์ปกติสำหรับการ merge
missing_data_merged = missing_per_station_filtered.reset_index().merge(
    total_records_per_station.reset_index(),
    on='tele_station_oldcode',
    how='left'
)

# 3. คำนวณสัดส่วนข้อมูลที่หายไป
missing_data_merged['proportion_rainfall_missing'] = missing_data_merged['rainfall_value'] / missing_data_merged['total_records']
missing_data_merged['proportion_tele_station_id_missing'] = missing_data_merged['tele_station_id'] / missing_data_merged['total_records']

# จัดรูปแบบผลลัพธ์ให้เป็นสถานีและสัดส่วน
missing_proportions = missing_data_merged[['tele_station_oldcode', 'proportion_rainfall_missing', 'proportion_tele_station_id_missing']]

if not missing_proportions.empty:
    print("สัดส่วนข้อมูลที่สูญหายไปของแต่ละสถานี (แสดงเฉพาะสถานีที่มีข้อมูลหาย):")
    # จัดเรียงตามสัดส่วนที่หายไปมากที่สุด
    display(missing_proportions.sort_values(by=['proportion_rainfall_missing', 'proportion_tele_station_id_missing'], ascending=False).round(4))
else:
    print("ไม่พบข้อมูลที่สูญหายไปในสถานีใดๆ ในคอลัมน์ 'rainfall_value' หรือ 'tele_station_id'")
```

# ข้อ 4 คัดสถานีที่มีข้อมูลหาย >= 20%

```python
# กรองสถานีที่มีสัดส่วนข้อมูล 'rainfall_value' หรือ 'tele_station_id' สูญหาย >= 20% (0.20)
stations_with_high_missing_data = missing_proportions[
    (missing_proportions['proportion_rainfall_missing'] >= 0.20) |
    (missing_proportions['proportion_tele_station_id_missing'] >= 0.20)
]

if not stations_with_high_missing_data.empty:
    print("สถานีที่มีข้อมูลสูญหาย >= 20% ในคอลัมน์ 'rainfall_value' หรือ 'tele_station_id':")
    display(stations_with_high_missing_data.sort_values(by=['proportion_rainfall_missing', 'proportion_tele_station_id_missing'], ascending=False).round(4))
    # สามารถเก็บรายชื่อสถานีเหล่านี้ไว้ในตัวแปรเพื่อใช้ในการกรอง DataFrame หลักในขั้นตอนต่อไปได้
    stations_to_exclude = stations_with_high_missing_data['tele_station_oldcode'].tolist()
    print("\nรายชื่อสถานีที่ควรถูกคัดออก:", stations_to_exclude)
else:
    print("ไม่พบสถานีใดๆ ที่มีข้อมูลสูญหายตั้งแต่ 20% ขึ้นไปในคอลัมน์ 'rainfall_value' หรือ 'tele_station_id'")
```

## สร้างคอลัมน์เดือนและปี

### Subtask:
แปลงคอลัมน์ `measure_datetime` ใน `filtered_combined_df` ให้เป็นรูปแบบ Datetime ถ้ายังไม่ได้ทำ แล้วสร้างคอลัมน์ใหม่ `month` และ `year` โดยดึงค่ามาจาก `measure_datetime`


**Reasoning**:
The 'measure_datetime' column has already been converted to datetime type in a previous step. I will now extract the month and year from this column to create new 'month' and 'year' columns in the 'filtered_combined_df' DataFrame as requested by the subtask instructions.



```python
filtered_combined_df['month'] = filtered_combined_df['measure_datetime'].dt.month
filtered_combined_df['year'] = filtered_combined_df['measure_datetime'].dt.year

print("สร้างคอลัมน์ 'month' และ 'year' จากคอลัมน์ 'measure_datetime' สำเร็จแล้ว")
display(filtered_combined_df.head())
```

**Reasoning**:
The previous code produced a `SettingWithCopyWarning`. Although the columns were added, this warning indicates potential issues when modifying a DataFrame that is a slice of another DataFrame. To resolve this, I will explicitly create a copy of `filtered_combined_df` before adding new columns, ensuring that modifications are made on an independent DataFrame.



```python
filtered_combined_df = filtered_combined_df.copy()
filtered_combined_df['month'] = filtered_combined_df['measure_datetime'].dt.month
filtered_combined_df['year'] = filtered_combined_df['measure_datetime'].dt.year

print("สร้างคอลัมน์ 'month' และ 'year' จากคอลัมน์ 'measure_datetime' สำเร็จแล้ว")
display(filtered_combined_df.head())
```

```python
monthly_rainfall = filtered_combined_df.groupby(['station_code', 'year', 'month'])['rainfall_value'].sum().reset_index()
monthly_rainfall.rename(columns={'station_code': 'Node', 'rainfall_value': 'rain'}, inplace=True)

print("คำนวณปริมาณน้ำฝนรายเดือนเสร็จสมบูรณ์:")
display(monthly_rainfall.head())
```
